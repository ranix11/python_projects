# Scatter plots to check the data

```python
sns.scatterplot(x= 'TotalPositiveCases',y='Recovered',data=data)
# this means that people are recovering as the number of cases increasing
```

```python
# total cases and how many people have died
sns.scatterplot(x= 'TotalPositiveCases',y='Deaths',data=data)
```

```python
# check the relationship between total cases and CurrentPositiveCases 
sns.scatterplot(x= 'TotalPositiveCases',y='CurrentPositiveCases',data=data)
```

```python
# relationship between HospitalizedPatients and total cases
sns.scatterplot(x= 'TotalPositiveCases',y='HospitalizedPatients',hue = 'Recovered',data=data)
```
